```python
import pandas as pd
```

```python
automobile = pd.read_csv("automobile.csv")
```

```python
automobile.head()
```

```python
import matplotlib.pyplot as plt
```

```python
plt.scatter(automobile["weight"], automobile["mpg"])
plt.xlabel("Peso del automóvil")
plt.ylabel("Consumo de combustible (mpg)")
plt.title("Relación entre peso y consumo")
plt.show()
```

```python
plt.hist(automobile["mpg"], bins=15)
plt.xlabel("Consumo de combustible (mpg)")
plt.ylabel("Cantidad de automóviles")
plt.title("Distribución del consumo de combustible")
plt.show()
```

```python
plt.boxplot([automobile["cylinders"], automobile["origin"]])
plt.xticks([1, 2], ["Cilindros", "Origen"])
plt.ylabel("Valor")
plt.title("Boxplot de cilindros y origen")
plt.show()
```

```python
automobile[["cylinders", "origin"]].dtypes
```

```python
automobile["origin"] = pd.to_numeric(automobile["origin"])
```

```python
automobile["origin"].unique()
```

```python
origen_num = automobile["origin"].map({"usa": 1, "europe": 2, "japan": 3})

plt.boxplot([automobile["cylinders"], origen_num])
plt.xticks([1, 2], ["Cilindros", "Origen"])
plt.ylabel("Valor")
plt.title("Boxplot de cilindros y origen")
plt.show()
```

```python
datos_barras = automobile.groupby(["model_year", "origin"]).size().unstack(fill_value=0)

datos_barras.plot(kind="bar", figsize=(12, 6))

plt.xlabel("Año del modelo")
plt.ylabel("Cantidad de automóviles")
plt.title("Distribución de automóviles por año y origen")
plt.legend(title="Origen")
plt.show()
```
